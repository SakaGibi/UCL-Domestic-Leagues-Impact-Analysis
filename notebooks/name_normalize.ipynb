```python
import pandas as pd
import os
```

```python
def load_league(path, competition_name):
    dfs = []

    for file in os.listdir(path):
        if file.endswith('.csv'):
            df = pd.read_csv(os.path.join(path, file))

            # season parse
            # ex: serie_a_20-21.csv
            season_part = file.split('_')[1]
            start, end = season_part.split('-')
            season = f"20{start}-20{end}"

            df['season'] = season
            df['competition'] = competition_name

            dfs.append(df)
    
    return pd.concat(dfs, ignore_index=True)

# Load all matches dataset
all_matches = pd.read_csv("../data/processed/all_matches_raw.csv")

# Load all leagues
serie_a = load_league("../data/raw/serie-a", "Serie_A")
premier_league = load_league("../data/raw/premier-league", "Premier_League")
bundesliga = load_league("../data/raw/bundesliga", "Bundesliga")
la_liga = load_league("../data/raw/la-liga", "La_Liga")
ligue_1 = load_league("../data/raw/ligue-1", "Ligue_1")
super_lig = load_league("../data/raw/super-lig", "Super_Lig")
champions_league = load_league("../data/raw/champions-league", "Champions_League")

# unique teams in each league
print(f"Serie A: {(serie_a['Home Team'].unique())}")
print(f"Premier League: {(premier_league['Home Team'].unique())}")
print(f"Bundesliga: {(bundesliga['Home Team'].unique())}")
print(f"La Liga: {(la_liga['Home Team'].unique())}")
print(f"Ligue 1: {(ligue_1['Home Team'].unique())}")
print(f"Super Lig: {(super_lig['Home Team'].unique())}")
print(f"Champions League: {(champions_league['Home Team'].unique())}")

# unique teams in all leagues
print(f"Total: {len(serie_a['Home Team'].unique()) + len(premier_league['Home Team'].unique()) + len(bundesliga['Home Team'].unique()) + len(la_liga['Home Team'].unique()) + len(ligue_1['Home Team'].unique()) + len(super_lig['Home Team'].unique()) + len(champions_league['Home Team'].unique())}")
```
